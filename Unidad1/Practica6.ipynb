```python
sns.set_style("whitegrid")

sns.boxplot(x="body-style", y="price", data=autos, hue="body-style", legend=False)

plt.title("Relacion entre precio y tipo de vehiculo", fontsize=14, fontweight="bold")
plt.xlabel("Tipo de vehiculo", fontsize=12)
plt.ylabel("Precio", fontsize=12)
plt.show()
```

**Interpretación:**
**Segmento accesible:** Los vehículos estilo hatchback presentan la mediana más baja y una dispersión reducida, consolidándose claramente como la opción de carrocería más económica del mercado.

**Versatilidad del mercado:** La categoría sedan muestra una amplitud enorme, cubriendo desde precios bajos hasta registrar múltiples valores atípicos (outliers) por encima de los 30,000. Esto refleja que existen sedanes tanto utilitarios como de gran lujo.

**Vehículos premium:** Los estilos hardtop y convertible registran las medianas y rangos de precios más altos, además de una enorme dispersión, indicando que son vehículos dirigidos a un consumidor con mayor poder adquisitivo.

```python
sns.boxplot(x="drive-wheels", y="price", data=autos, hue="drive-wheels", legend=False)

plt.title("Relacion entre precio y tipo de traccion", fontsize=14, fontweight="bold")
plt.xlabel("Tipo de traccion", fontsize=12)
plt.ylabel("Precio", fontsize=12)
plt.show()
```

**Interpretación:**

**Tracción económica:** Los vehículos con tracción delantera (fwd) son los más baratos y tienen la mayor concentración de sus datos en precios bajos, congruente con la fabricación en masa de autos de uso cotidiano y citadino.

**Tracción de alto costo:** La tracción trasera (rwd) posee la mediana más alta y un rango de precios muy extendido. Este tipo de tracción suele asociarse a vehículos deportivos o de lujo, justificando su elevado impacto en el costo final.

*Punto intermedio:** La tracción en las cuatro ruedas (4wd) se mantiene en un rango estrecho e intermedio. Esto suele representar SUVs o vehículos utilitarios de gama media que no llegan a los extremos de precios de las otras configuraciones.
