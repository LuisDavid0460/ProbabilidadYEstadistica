```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [0, 10, 20, 50]         # Valores posibles (Monto en pesos)
p = [0.50, 0.30, 0.15, 0.05]  # Probabilidades correspondientes

esperanza = 0
for i in range(len(x)):
    esperanza += x[i] * p[i]

esperanza_cuadrado = 0
for i in range(len(x)):
    esperanza_cuadrado += (x[i] ** 2) * p[i]
```

```python
varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)

print(f"Suma de p = {sum(p):.2f}\n")

print("3. Calcule:")
print(f"E(X) = {esperanza:.2f}")
print(f"E(X^2) = {esperanza_cuadrado:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviación estándar = {desviacion:.2f}")
```

```python
p_array = np.array(p)
F = np.cumsum(p_array)

for xi, fi in zip(x, F):
    print(f"F({xi}) = {fi:.2f}")
```

```python
plt.step(x, F, where='post')
plt.title("Función de Distribución Acumulada (FDA) - Tómbola")
plt.xlabel("Monto del premio (Pesos)")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()
```

```python
x_arr = np.array(x)
p_arr = np.array(p)

plt.stem(x_arr, p_arr, basefmt=" ", linefmt='b-', markerfmt='bo')
plt.title("Distribución de Probabilidad Discreta\ncon Media y Desviación Estándar")
plt.xlabel("Monto del Premio en pesos (X)")
plt.ylabel("P(X)")
plt.grid(True, linestyle='--', alpha=0.6)

plt.axvline(x=esperanza, color='red', linestyle='--', label=f"Media (E[X]) = {esperanza:.2f}")

plt.axvspan(esperanza - desviacion, esperanza + desviacion, color='red', alpha=0.2,
            label=f"±1 Desv. Estándar (σ = {desviacion:.2f})")

plt.legend()
plt.ylim(0, max(p_arr) + 0.05)
plt.xticks(x_arr)

plt.show()
```

# **Interpretacion de los datos**

## **¿Que significa el valor esperado en este contexto?**

### El valor esperado (media) es de 8.50 pesos. En este contexto, significa que si una persona jugara a la tombola muchisimas veces, la "ganancia promedio" a largo plazo que obtendria por cada boleto que saque seria de 8.50 pesos

## **¿El juego es favorable para el jugador?**

### Tiene 3 opciones

### Si el boleto cuesta mas de 8.50 pesos, entonces no seria favorable para el jugador
### Si el boleto cuesta menos de 8.50 pesos, entonces si seria favorable para el jugador
