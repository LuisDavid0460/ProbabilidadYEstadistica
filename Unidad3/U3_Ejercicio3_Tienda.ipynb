```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [0, 1, 2, 3, 4, 5]                 # Valores posibles (Unidades)
p = [0.05, 0.10, 0.25, 0.30, 0.20, 0.10] # Probabilidades correspondientes

esperanza = 0
for i in range(len(x)):
    esperanza += x[i] * p[i]
esperanza_cuadrado = 0
for i in range(len(x)):
    esperanza_cuadrado += (x[i] ** 2) * p[i]
```

```python
varianza = esperanza_cuadrado - esperanza**2
desviacion = math.sqrt(varianza)
print("1. Verifique que las probabilidades sumen 1:")
print(f"Suma de p = {sum(p):.2f}\n")

print("3. Calcule:")
print(f"E(X) = {esperanza:.2f}")
print(f"E(X^2) = {esperanza_cuadrado:.2f}")
print(f"Varianza = {varianza:.2f}")
print(f"Desviación estándar = {desviacion:.2f}")
```

```python
p_array = np.array(p)
F = np.cumsum(p_array)

print("\n4. Construya la función de distribución acumulada F(x):")
for xi, fi in zip(x, F):
    print(f"F({xi}) = {fi:.2f}")
```

```python
plt.step(x, F, where='post')
plt.title("Función de Distribución Acumulada (FDA) - Ventas")
plt.xlabel("Unidades vendidas (X)")
plt.ylabel("Probabilidad acumulada")
plt.grid(True)
plt.ylim(0, 1.05)
plt.show()
```

```python
x_arr = np.array(x)
p_arr = np.array(p)
plt.stem(x_arr, p_arr, basefmt=" ", linefmt='b-', markerfmt='bo')
plt.title("Distribución de Probabilidad Discreta\ncon Media y Desviación Estándar")
plt.xlabel("Unidades vendidas por semana (X)")
plt.ylabel("P(X)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.axvline(x=esperanza, color='red', linestyle='--', label=f"Media (E[X]) = {esperanza:.2f}")
plt.axvspan(esperanza - desviacion, esperanza + desviacion, color='red', alpha=0.2,
            label=f"±1 Desv. Estándar (σ = {desviacion:.2f})")


plt.legend()
plt.ylim(0, max(p_arr) + 0.05)
plt.xticks(x_arr)
plt.show()
```
