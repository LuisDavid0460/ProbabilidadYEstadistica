# **GRUPO 1**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

# 1. Carga de Datos
df1 = pd.read_csv("grupo1.csv")

# 2. Exploración Inicial (Descripción estadística)
print("--- Estadísticas Descriptivas Grupo 1 ---")
print(df1.describe())

# 3. Generar Histograma de Calificaciones
plt.figure(figsize=(6, 4))
plt.hist(df1['Calificacion'], bins=10, color='skyblue', edgecolor='black')
plt.title('Histograma de Calificaciones')
plt.xlabel('Calificación')
plt.ylabel('Frecuencia')
plt.show()

# 4. Entrenar el Modelo de Regresión Lineal
X = df1[['HorasEstudio']]
Y = df1['Calificacion']

modelo = LinearRegression()
modelo.fit(X, Y)

# 5. Obtener Ecuación (Pendiente e Intercepto)
pendiente = modelo.coef_[0]
intercepto = modelo.intercept_
print(f"Ecuación de la recta: Y = {intercepto:.2f} + {pendiente:.2f}X")

# 6. Calcular Coeficiente de Correlación de Pearson
correlacion = df1['HorasEstudio'].corr(df1['Calificacion'])
print(f"Coeficiente de correlación de Pearson: {correlacion:.4f}")
```

```python
# Diagrama de dispersion con linea de regresion
plt.figure(figsize=(6, 4))
sns.set(style="whitegrid")
sns.regplot(x='HorasEstudio', y='Calificacion', data=df1, scatter=True, color='blue')
plt.title('Relacion entre Horas de estudio y Califacacion final')
plt.xlabel('Horas de estudio')
plt.ylabel('Califacion final')
plt.show()
```

# **RESULTADOS - GRUPO 1**

### El coeficiente de regresion es de -0.11. Esto significa que por cada hora adicional de estudio, la califacacion final tiende a disminuir 0.11 puntos, lo cual es un impacto minimo
### El coeficiente de correlacion de Pearson es de -0.0571. Esto indica que practicamente no existe ninguna relacion lineal entre el tiempo de estudio y las califaciones en este grupo, por lo que la relacion es muy debil
### El modelo no es estadisticamente significativo. En un diagrama de dispersion, los puntos estan completamente dispersos sin seguir el patron de la linea de regresion
### El modelo es nada confiable. El coeficiente de determinarion es de apenas 0.0033, lo que significa que el tiempo de estudio explica menos del 1% de las variaciones en las calificaciones finales. No se puede predecir el rendimiento de estos alumnos basandose en sus horas de estudio
