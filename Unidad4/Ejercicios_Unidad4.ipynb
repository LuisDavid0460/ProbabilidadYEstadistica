```python
from scipy.stats import poisson

# Parametros de la distribucion de Poisson
media_accidentes = 2

# Parte 1: Probabilidad de que no haya accidentes en 1 semana
probabilidad_0_accidentes = poisson.pmf(0, media_accidentes)

print("1. Probabilidad de que no haya accidentes en 1 semana:", probabilidad_0_accidentes)

# Parte 2: Probabilidad de que a lo sumo haya tres accidentes en 2 semanas
# Convertimos la media a una media para 2 semanas
media_2_semanas = media_accidentes * 2

# Calculamos la probabilidad acumulada hasta 3 accidentes en 2 semanas
probabilidad_3_o_menos_accidentes = poisson.cdf(3, media_2_semanas)

print("2. Probabilidad de que a lo sumo haya tres accidentes en 2 semanas:", probabilidad_3_o_menos_accidentes)
```

# **EJERCICIO 1**. Peso de 500 estudiantes

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

plt.figure(figsize=(6,4))

media = 70
desv_std = 4
total_estudiantes = 500

peso_limite_inferior = 68
peso_limite_superior = 72

prob_hasta_72kg = norm.cdf(peso_limite_superior, media, desv_std)
prob_hasta_68kg = norm.cdf(peso_limite_inferior, media, desv_std)

prob_entre_68_y_72kg = prob_hasta_72kg - prob_hasta_68kg

estudiantes_entre_68_y_72 = total_estudiantes * prob_entre_68_y_72kg

print("Probabilidad de que un esstudiante pese entre 68 y 72 kg:", prob_entre_68_y_72kg)
print("Número de estudiantes que pesarán entre 68 y 72 kg:", estudiantes_entre_68_y_72)

x = np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y = norm.pdf(x, media, desv_std)

plt.plot(x, y, 'b', linewidth=2)

plt.fill_between(x, y, where=((x >= peso_limite_inferior) & (x <= peso_limite_superior)), color='skyblue', alpha=0.5)
plt.title('Probabilidad de que un estudiante pese entre 68 y 72 kg. \n $\mu=70$, $\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```

# **EJERCICIO 2**. Problema 1 Distribuccion binomial y poisson

## Problema: El numero x de personas ingresadas a una unidad de cuidados intensuivos en un hospoital particual, en un dia, tiene una distribucion de probabilidad de Poisson con media igual a cinco personas por dia.
## A) ¿Cual es la probabilidad de que el numero de personas ingresadas a una unidad de cuidados intensivos en un hospital particular, en un dia particular, sea dos?
## B) ¿Es probable que x exceda de 10? Explicar

### **Explicacion**: El tipo de distribucion es **Poisson**, el problema establece explicitamente que la variable tiene una "distribucionm de probabilidad de Poisson". Ademas, este modelo es el adecuado porque cuenta el numero de eventos que ocurren en un intervalo de tiempo con una tasa media conocida.

```python
from scipy.stats import poisson

# Parámetros de la distribución de Poisson
media_ingresos = 5 # Media de personas por día

# a) ¿Cuál es la probabilidad de que el número de personas ingresadas sea dos?
x_a = 2
probabilidad_2_personas = poisson.pmf(x_a, media_ingresos)
print("a) La probabilidad de que ingresen exactamente 2 personas es:", probabilidad_2_personas)

# b) ¿Es probable que x exceda de 10? Explicar
x_b = 10
# Calculamos la probabilidad acumulada hasta 10 y se la restamos a 1 para obtener P(X > 10)
probabilidad_excede_10 = 1 - poisson.cdf(x_b, media_ingresos)
print("b) La probabilidad de que exceda de 10 personas es:", probabilidad_excede_10)
```

### Respuesta a la pregunta b): No es probable

# **EJERCICIO 3**. Problema 2 Distribuccion binomial y poisson

## Problema: Realizar un programa en python que resuelva el siguiente problema:

## - Un fabricante de podadoras para el pasto compra motores de 1 hp y 2 ciclos, en lotes de 1000, a un proveedor.
## - El fabricante equipa cada una de las podadoras producidas por su planta con uno de los motres.
## - La historia muestra que la probabilidad de que cualquier motor del proveedor resulte no satisfactoria es .001.
## - **En un embarque de 1000 motores, calcular**
## - ¿Cual es la probabilidad de que ninguno sea defectuoso? , ¿Cual es la probabilidad de que tres sean defectuosos? , ¿Cual es la probabilidad de que cuatro sean defectuosos?

### **Explicacion**:
### El tipo de distribucionm es **Binomial**, se trata de un numero fijo de ensayos independientes, donde cada ensayo tiene solo dos resultados posibles, si es defectuoso o no. Y la probabilidad de exito es constante


```python
from scipy.stats import binom

# Parámetros de la distribución binomial
n = 1000 # Lote de motores
p = 0.001 # Probabilidad de que un motor sea defectuoso

# ¿cuál es la probabilidad de que ninguno sea defectuoso?
x_0 = 0
probabilidad_ninguno = binom.pmf(x_0, n, p)
print("La probabilidad de que ninguno sea defectuoso es:", probabilidad_ninguno)

# ¿cuál es la probabilidad de que tres sean defectuosos?
x_3 = 3
probabilidad_tres = binom.pmf(x_3, n, p)
print("La probabilidad de que tres sean defectuosos es:", probabilidad_tres)

# ¿cuál es la probabilidad de que cuatro sean defectuoso?
x_4 = 4
probabilidad_cuatro = binom.pmf(x_4, n, p)
print("La probabilidad de que cuatro sean defectuosos es:", probabilidad_cuatro)
```
